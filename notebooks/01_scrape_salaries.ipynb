# 01 — Scrape NBA Salaries (ESPN), 2024-25

Scrapes the 13 pages of ESPN's NBA salary table and saves the raw strings (no conversion) to `data/raw/nba_salaries_2024-25.xlsx`.

The scraping logic lives in `src/nba_roi/scrape_salaries.py`; the same step can be run from the command line with `python scripts/fetch_data.py --salaries`.

```python
from nba_roi import config
from nba_roi.scrape_salaries import scrape_espn_salaries

salaries = scrape_espn_salaries(year=config.ESPN_SALARY_YEAR, pages=config.ESPN_SALARY_PAGES)
print(f"Rows scraped: {len(salaries)}")
salaries.head()
```

ESPN repeats its header row on every page (`NAME / TEAM / SALARY`). Those rows are kept in the raw file and removed during cleaning in notebook 03.

```python
config.SALARIES_FILE.parent.mkdir(parents=True, exist_ok=True)
salaries.to_excel(config.SALARIES_FILE, index=False)
print(f"Saved -> {config.SALARIES_FILE.relative_to(config.PROJECT_ROOT)}")
```
